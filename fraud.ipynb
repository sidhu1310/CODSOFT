```python

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

data = pd.read_csv("/kaggle/input/fraud-detection/fraudTrain.csv")

categorical_columns = ['merchant', 'category', 'gender', 'job', 'state']
numeric_columns = ['amt', 'city_pop', 'unix_time', 'merch_lat', 'merch_long', 'lat', 'long']

drop_columns = ['Unnamed: 0', 'trans_date_trans_time', 'cc_num', 'first', 'last', 'street', 'city', 'zip', 'dob', 'trans_num']

encoder = OneHotEncoder(sparse=True, handle_unknown='ignore')
X_encoded = encoder.fit_transform(data[categorical_columns])

X_numeric = data[numeric_columns]

X_encoded_df = pd.DataFrame.sparse.from_spmatrix(X_encoded, columns=encoder.get_feature_names_out(categorical_columns))
X_combined = pd.concat([X_numeric.reset_index(drop=True), X_encoded_df.reset_index(drop=True)], axis=1)

y = data['is_fraud']

X_combined.head(), y.head()
```

```python
from sklearn import tree
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
xtrain,ytrain,xtest,ytest = train_test_split(X_combined,y,test_size=0.2,random_state=42)
from sklearn.preprocessing import LabelEncoder
```

```python
# Check the shapes before splitting
print(f"X_combined shape: {X_combined.shape}")
print(f"y shape: {y.shape}")

# Check the shapes after splitting
print(f"xtrain shape: {xtrain.shape}")
print(f"ytrain shape: {ytrain.shape}")
print(f"xtest shape: {xtest.shape}")
print(f"ytest shape: {ytest.shape}")

# Check the type of y
print(f"Type of y: {type(y)}")
```

```python

clf = tree.DecisionTreeClassifier()

clf.fit(xtrain, xtest)
```

```python
res = clf.predict(ytrain)
accuracy_score(res,ytest)
```
